```python
def perceptron(x1, x2, w1, w2, bias):
    linear_output = (x1 * w1) + (x2 * w2) + bias
    return 1 if linear_output >= 0 else 0

# AND Gate implementation
def AND_gate(x1, x2):
    return perceptron(x1, x2, w1=1, w2=1, bias=-1.5)

print(f"AND(0,0): {AND_gate(0,0)}") # Expected output: 0
print(f"AND(1,0): {AND_gate(1,0)}") # Expected output: 0
print(f"AND(0,1): {AND_gate(0,1)}") # Expected output: 0
print(f"AND(1,1): {AND_gate(1,1)}") # Expected output: 1
```

```python
def OR_gate(x1, x2):
    return perceptron(x1, x2, w1=1, w2=1, bias=-0.5)

def NAND_gate(x1, x2):
    return perceptron(x1, x2, w1=-1, w2=-1, bias=1.5)

def XOR_gate(x1, x2):
    # Layer 1
    gate1 = OR_gate(x1, x2)
    gate2 = NAND_gate(x1, x2)

    # Layer 2 (Output)
    return AND_gate(gate1, gate2)

print(f"XOR(0,0): {XOR_gate(0,0)}") # Expected output: 0
print(f"XOR(1,0): {XOR_gate(1,0)}") # Expected output: 1
print(f"XOR(0,1): {XOR_gate(0,1)}") # Expected output: 1
print(f"XOR(1,1): {XOR_gate(1,1)}") # Expected output: 0
```
